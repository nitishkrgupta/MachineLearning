```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
data = {
    "Color": [
        "Red", "Blue", "Green", "Red", "Blue",
        "Green", "Yellow", "Black", "White", "Yellow"
    ],
    "Fruit": [
        "Apple", "Mango", "Grapes", "Cherry", "Orange",
        "Kiwi", "Banana", "Jamun", "Coconut", "Pineapple"
    ]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
x = df[['Color']]
y = df['Fruit']
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x, y, train_size=0.8, random_state=42)
```

```python
label = LabelEncoder()
```

```python
label.fit_transform(xtrain['Color'])
```

```python
label.transform(xtest['Color'])
```

### Conclusion

after spliting the data there may be a chance of values present in train which will not present in the test or the values present in the test which will not present in the train.
after fit_transform it learns the pattern from the values but it will not learn the pattern for those values which is present in the test data. because of this it throws error.
