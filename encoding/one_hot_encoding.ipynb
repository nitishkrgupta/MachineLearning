```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
```

```python
data = {
    "Outlook": ["Sunny", "Rainy", "Cloudy", "Sunny", "Rainy",
                "Cloudy", "Sunny", "Rainy", "Cloudy", "Sunny"],

    "Temperature": ["Hot", "Cold", "Mild", "Hot", "Cold",
                    "Mild", "Hot", "Cold", "Mild", "Hot"],

    "Wind": ["Strong", "Weak", "Strong", "Weak", "Strong",
             "Weak", "Strong", "Weak", "Strong", "Weak"],

    "Play": ["No", "Yes", "Yes", "No", "Yes",
             "Yes", "No", "Yes", "Yes", "No"]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
x = df.drop(columns='Play')
y = df['Play']
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x, y, train_size=0.7, random_state=42)
```

```python
one_hot_encode = OneHotEncoder()
```

```python
one_hot_encode.fit_transform(xtrain[['Outlook', 'Temperature', 'Wind']])
```
