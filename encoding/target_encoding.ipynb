```python
import pandas as pd
from category_encoders import TargetEncoder
from sklearn.model_selection import train_test_split
```

### Scenario - 1

when target column has numercal continuous values

```python
data = {
    "Education": [
        "High School", "Bachelor", "Master", "PhD",
        "Bachelor", "Master", "High School", "PhD",
        "Master", "Bachelor"
    ],

    "Experience": [
        "Fresher", "Junior", "Mid", "Senior",
        "Lead", "Mid", "Junior", "Lead",
        "Senior", "Fresher"
    ],

    "Salary": [
        20000, 25000, 40000, 50000, 45000, 40000, 25000, 35000, 50000, 20000
    ]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
x = df.drop(columns='Salary')
y = df['Salary']
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x, y, train_size=0.8, random_state=42)
```

```python
encoder = TargetEncoder()

encoder.fit_transform(xtrain, ytrain)
```

### Scenario - 2

when target column has categorical values

```python
data = {
    "Education": [
        "High School", "Bachelor", "Master", "PhD",
        "Bachelor", "Master", "High School", "PhD",
        "Master", "Bachelor"
    ],

    "Experience": [
        "Fresher", "Junior", "Mid", "Senior",
        "Lead", "Mid", "Junior", "Lead",
        "Senior", "Fresher"
    ],

    "Performance": [
        "Poor", "Average", "Good", "Excellent",
        "Excellent", "Good", "Average", "Poor",
        "Good", "Excellent"
    ],

    "Salary": [
        "Low", "Medium", "High", "Very High",
        "Very High", "High", "Medium", "Low",
        "High", "Medium"
    ]
}
```

```python
df1 = pd.DataFrame(data)
df1
```

```python
x = df1.drop(columns='Salary')
y = df1['Salary']
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x, y, train_size=0.8, random_state=42)
```

```python
encoder = TargetEncoder()

encoder.fit_transform(xtrain, ytrain)
```

```python
encoder.transform(xtest, ytest)
```

```python
encoder.transform(xtest)
```

```python
encoder = TargetEncoder(cols=['Education'])

encoder.fit_transform(xtrain, ytrain)
```
