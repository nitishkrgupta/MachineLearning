```python
import pandas as pd
from sklearn.ensemble import IsolationForest
```

```python
df = pd.read_csv('creditcard.csv')
df.head()
```

```python
df.shape
```

```python
model = IsolationForest(n_estimators=100, 
                        max_samples=0.8, 
                        max_features=1.0, 
                        bootstrap=True, 
                        n_jobs=8, 
                        verbose=1,
                        contamination=0.05) # 5% anomaly values to detect
```

```python
result = model.fit_predict(df)
```

```python
anomalies = df[result==-1]
normal = df[result==1]
```
