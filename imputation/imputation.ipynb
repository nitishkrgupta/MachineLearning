```python
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
data = {
    "Age": [25, 30, None, 45, None],
    "Salary": [50000, None, 55000, None, 52000],
    "City": ["Manglore", "Bangalore", "Mumbai", None, "Delhi"]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
imputer = SimpleImputer(strategy='mean')
imputer.fit_transform(df[['Age', 'Salary']])
```

```python
# imputer = SimpleImputer(strategy='mean')
# imputer.fit_transform(df[['City']])
```

```python
imputer = SimpleImputer(strategy='most_frequent')
imputer.fit_transform(df[['Age']])
```

```python
imputer = SimpleImputer(strategy='most_frequent')
imputer.fit_transform(df[['City']])
```

```python
imputer = SimpleImputer(strategy="constant", fill_value=0)
```

```python
imputer.fit_transform(df[['Age', 'Salary']])
```
